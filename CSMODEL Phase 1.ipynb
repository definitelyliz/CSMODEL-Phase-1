## Dataset Description



#### Brief description of the dataset
The Family Income and Expenditure Survey 2012 (FIES) is a nationwide survey conducted by the Philippine Statistics Authority (PSA) to collect data on income and expenditures of Filipino households. The data from this survey is used to assess the socio-economic conditions of families across 17 regions in the Philippines. The survey provides vital information that aids in the formulation of government policies, economic plans, and poverty alleviation programs.

#### Collection process and implications
The FIES 2012 used a two-phase collection process, gathering data in July 2012 for the first round and in January 2013 for the second round. The survey covered a national sample of around 50,000 households and was conducted via face-to-face interviews with household members. The survey focused on detailed questions about household income, sources of income, and various types of expenditures (e.g., food, housing, education).
Implications of the data include its use in shaping national and regional development programs, economic forecasts, and understanding the cost of living across various demographics. Furthermore, it provides insights into income inequality and poverty levels across different regions and social classes.

#### Dataset Structure
- The dataset contains information from around 50,000 households nationwide.
- The dataset is structured into 17 geographic regions based on the Philippines' administrative division.
- Each observation represents a household, with the primary unit of analysis being the household head and its members.

#### Dataset Variables
The FIES 2012 dataset includes the following main variables:
1. **Household Income**: Total household income from all sources (e.g., wages, agricultural income, remittances).
2. **Household Expenditure**: Includes expenditures on food, clothing, housing, education, health, and other services.
3. **Demographic Information**: Data on household composition, including the number of family members, ages, employment status, and educational background.
4. **Geographic Variables**: Indicators for the administrative region, province, and barangay (village level).
5. **Other Income Sources**: This includes income from non-work sources such as pensions, government assistance, and financial investments.

Fix incorrect data types for EMPLOYED_PAY and EMPLOYED_PROF and compute for their average.

```python
newDf = pd.read_csv("filtered_file.csv")

# Compute for average and set data type to integer
def average(size, columnName):
    blankCounter = 0
    ave = 0
    for i in range(size):
        if(newDf[columnName][i] == "  "):
            blankCounter += 1
        else:
            newDf.loc[i, columnName] = int(newDf.loc[i, columnName])
            ave = ((ave * (i - blankCounter)) + (newDf.loc[i, columnName])) / (i + 1 - blankCounter)
    return int(round(ave))

employedPayAve = average(newDf['EMPLOYED_PAY'].size, 'EMPLOYED_PAY')
employedProfAve = average(newDf['EMPLOYED_PROF'].size, 'EMPLOYED_PROF')

# Print each average
print(employedPayAve)
print(employedProfAve)
```
