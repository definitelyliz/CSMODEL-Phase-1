## Dataset Description



#### Brief description of the dataset
The Family Income and Expenditure Survey 2012 (FIES) is a nationwide survey conducted by the Philippine Statistics Authority (PSA) to collect data on income and expenditures of Filipino households. The data from this survey is used to assess the socio-economic conditions of families across 17 regions in the Philippines. The survey provides vital information that aids in the formulation of government policies, economic plans, and poverty alleviation programs.

#### Collection process and implications
The FIES 2012 used a two-phase collection process, gathering data in July 2012 for the first round and in January 2013 for the second round. The survey covered a national sample of around 50,000 households and was conducted via face-to-face interviews with household members. The survey focused on detailed questions about household income, sources of income, and various types of expenditures (e.g., food, housing, education).
Implications of the data include its use in shaping national and regional development programs, economic forecasts, and understanding the cost of living across various demographics. Furthermore, it provides insights into income inequality and poverty levels across different regions and social classes.
The two-round collection method reduces seasonal biases but still requires careful consideration when analyzing changes in income or expenditure between rounds.


#### Dataset Structure
- **Rows**: Each row in the dataset represents a household.
- **Columns**: Each column represents a specific variable, such as income, expenditures, or demographic details.
- **Number of Observations (Rows)**: 40171 households.
- **Number of Variables (Columns)**: 119 variables, representing detailed household-level income, expenditures, and demographic data.

#### Cleaning `Region`

When we used the `info()` function, we saw that the `Region` variable was in `int64` numerical format. Since the variable represents a category and not numerical data, we will convert it to a categorical data type. Correct data representation will help in further analysis.

The Philippines has 17 different regions. As such, we expect that the possible values are whole numbers from 1 to 17. We'll use the `unique()` function to list all the unique values for this variable.

```python
df['Region'].unique()
```

We can see that there are 17 different unique values for the `Region` variable. However, we can also observe that Region 4 is not represented, and there are two values that fall under the range of 1 to 17 (41 and 42). 

We followed the [Family Income and Expenditure 2012 Documentation](https://psada.psa.gov.ph/catalog/38/related-materials)(page 20) coding system used for the `Region` variable. 41 and 41 were used to represent Region IVA and Region IVB.

We create another dictionary to map the codes to their names.

```python
region_names = {
    1: 'I - Ilocos Region',
    2: 'II - Cagayan Valley',
    3: 'III - Central Luzon',
    5: 'V - Bicol Region',
    6: 'VI - Western Visayas',
    7: 'VII - Central Visayas',
    8: 'VIII - Eastern Visayas',
    9: 'IX - Zamboanga Peninsula',
    10: 'X - Northern Mindanao',
    11: 'XI - Davao Region',
    12: 'XII - SOCCSKSARGEN',
    13: 'NCR',
    14: 'CAR',
    15: 'ARMM',
    16: 'Caraga',
    41: 'IVA - CALABARZON',
    42: 'IVB - MIMAROPA'
}
```

We use the [`map()`](https://docs.python.org/3/library/functions.html#map) function to iterate over each observation to replace the values with the region names. 

```python
df['Region'] = df['Region'].map(region_names)
df['Region'].unique()
```

We convert the data type to `category` since the values are limited to a set of region names [(reference)](https://stackoverflow.com/questions/30601830/when-to-use-category-rather-than-object).

```python
df['Region'] = df['Region'].astype('category')
df.info()
```

#### Cleaning `Total Number of Family Members`
As previously mentioned, `Total Number of Family Members` is a `float64` type when we expect it to be an `int64` type since there can't be a fractional value for the number of family members. We'll further investigate and check the best approach to clean the data, whether to round down or up.

We'll create two copies of `df` using the `copy()` function.

```python
df_round_down = df.copy()
df_round_up = df.copy()
```

We'll use `np.floor()` to round down and `np.ceil()` to round up. 

```python
df_round_down['Total Number of Family Members'] = np.floor(df_round_down['Total Number of Family Members']).astype(int)
df_round_up['Total Number of Family Members'] = np.ceil(df_round_up['Total Number of Family Members']).astype(int)
```

We'll check their summary statistics using the `agg()` function and choose the best approach based on the data.

```python
df_round_down.agg({'Total Number of Family Members': ['mean', 'median', 'std']})
```

```python
df.agg({'Total Number of Family Members': ['mean', 'median', 'std']})
```

```python
df_round_up.agg({'Total Number of Family Members': ['mean', 'median', 'std']})
```

Based on the aggregated statistics, the mean and median from both approaches have the same absolute difference from the original data. The standard deviation shows that rounding down maintains the statistical integrity given its smaller value and closer value to the original value while rounding up had a bigger value, which could increase variability.

Thus, we will use `np.floor()` on the `Total Number of Family Members.`

```python
df['Total Number of Family Members'] = np.floor(df['Total Number of Family Members']).astype('int64')
```

## Exploratory Data Analysis

### EDA 1: How does the total food expenditure for food consumed at home compare to the total food expenditure for food regularly consumed outside of home across different regions?

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x)
plt.style.use('seaborn-v0_8-darkgrid')

# Read csv and get the number of columns and rows
newDf = pd.read_csv("final_fies.csv")

# Group by region and sum the expenditures
expenditure_summary = newDf.groupby('Region').sum()

# Calculate the total expenditure for home and outside
totalHomeExpenditure = expenditure_summary['Total Food Expenditure at Home']
totalOutsideExpenditure = expenditure_summary['Total Food Expenditure Outside']

# Compare the expenditures
data = {
    'Total Food Expenditure at Home': totalHomeExpenditure,
    'Total Food Expenditure Outside': totalOutsideExpenditure
}

comparison = pd.DataFrame(data)
average = comparison['Total Food Expenditure at Home'].sum() / 40171

tFoodHomeMCTAndMD = comparison.agg({'Total Food Expenditure at Home': ['mean', 'median', 'std', 'var']})
tFoodOutsideMCTAndMD = comparison.agg({'Total Food Expenditure Outside': ['mean', 'median', 'std', 'var']})
print("Total Food Expenditure at Home measures of central tendency and measures of dispersion :\n\n",
    tFoodHomeMCTAndMD)
print("\nTotal Food Expenditure Outside measures of central tendency and measures of dispersion :\n\n",
    tFoodOutsideMCTAndMD)

correlation = totalHomeExpenditure.corr(totalOutsideExpenditure)

print("\nCorrelation value = ", correlation.round(2))

print("\nThis figure shows a bar graph between the amount of food consumed at home and the amount of food consumed\n outside across different regions: ")

# Plotting the expenditures
plt.figure(figsize=(8, 4))
comparison.plot(kind='bar')
plt.title('Total Food Expenditure Comparison by Region')
plt.xlabel('Region')
plt.ylabel('Total Expenditure')
plt.legend(['Total Food Expenditure at Home', 'Total Food Expenditure Outside'])
plt.show()

print("\nThis histogram shows the distribution of the total food consumed at home and the total food consumed outside: ")

# Plotting the distributions
totalHomeExpenditure.hist(bins=30, label='Total Food Expenditure at Home', figsize=(30, 15))
totalOutsideExpenditure.hist(bins=30, label='Total Food Expenditure Outside', figsize=(30, 15))
plt.title('Distribution of Food Expenditures')
plt.xlabel('Expenditure')
plt.legend()
plt.show()

# print(comparison)
```

### EDA 2: Is there a significant difference in total annual income across different regions?

This EDA question focuses on the total annual income for each region in the Philippines. The variables involved are `Total Income` and `Region`.

We use the `agg()` function to get the mead, median, and standard deviation of annual income per region.

```python
df_2 = pd.read_csv("final_fies.csv")
```

```python
# Total income summary statistic
income = df_2.agg({'Total Income': ['mean', 'median', 'std']})
print(income)

# Total income per region summary statistic
income_per_region = df_2.groupby('Region').agg({'Total Income': ['mean', 'median', 'std']})
income_per_region.sort_values(('Total Income', 'mean'), ascending = False)
```

The data above shows that the average income for 2012 is 217,618.68 pesos, and the median is 140,125.00 pesos. With the mean higher than the median and a relatively large standard deviation, the data tells us that the data is skewed.

When we grouped the data per region, the results show that the annual income difference between the region with the highest average and lowest average is quite large. NCR is the region with the highest average, while ARMM is the region with the lowest mean.

We will use a barplot and a boxplot to better see the difference of average annual income across regions.

```python
plt.figure(figsize=(16, 5))
plt.hist(df_2['Total Income'], bins=500)
plt.title('Total Income')
plt.ticklabel_format(style='plain', axis='x')
plt.xlabel('Total Income')
plt.ylabel('Count')
plt.show()
```

The histogram shows that the `Total Income` is positively skewed. It is observed that the data is mostly falls between the 0 to 2,000,000 pesos mark and there are outliers present from 4,000,000 to 8,000,000.

```python
df_2.boxplot("Total Income", by="Region", figsize=(20,20))
plt.title('Total Income by Region')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=90)
plt.xlabel('Region')
plt.ylabel('Total Income')
plt.show()
```

The box plot confirms our conclusion from the numerical summaries. Both the histogram and the boxplot show that the `Total Income` variable is skewed. It can also be seen in the boxplot that there are several outliers from each region. However, because of the size of the dataset, the `Total Income` range, and the outliers, the output of the boxplot is smaller than we'd like.



### EDA 3: Is there a relationship between the household size and the total household income?


```python
import seaborn as sns

# Load the dataset
file_path = 'final_fies.csv'
newDf = pd.read_csv(file_path)

# Step 1: Numerical summaries for 'Total Number of Family Members'
household_size_summary = {
    'mean': newDf['Total Number of Family Members'].mean(),
    'median': newDf['Total Number of Family Members'].median(),
    'std': newDf['Total Number of Family Members'].std()
}

# Numerical summaries for 'Total Income'
total_income_summary = {
    'mean': newDf['Total Income'].mean(),
    'median': newDf['Total Income'].median(),
    'std': newDf['Total Income'].std()
}

# Display the summaries
household_size_df = pd.DataFrame(household_size_summary, index=['Total Number of Family Members'])
income_df = pd.DataFrame(total_income_summary, index=['Total Income'])

print("\nNumerical Summary for Total Number of Family Members:")
display(household_size_df)

print("\nNumerical Summary for Total Income:")
display(income_df)

#Correlation between 'Total Number of Family Members' and 'Total Income'
correlation = newDf['Total Number of Family Members'].corr(newDf['Total Income'])
print(f"\nCorrelation between Total Number of Family Members and Total Income: {correlation:.2f}")
```

### Interpretation
Total Number of Family Members (Household Size):
- Mean: The average household size is about 4.58 members.
- Median: The median household size is 4 members, meaning half of the households have fewer than 4 members and the other half have more than 4 members.
- Standard Deviation: The spread or variability in household size is about 2.18 members.

Total Income:
- Mean: The average total household income is approximately ₱217,618.
- Median: The median household income is ₱140,125, indicating that most households earn below the mean (since the median is lower).
- Standard Deviation: The standard deviation of total income is quite high, at ₱253,695, indicating that there is significant variation in income across the dataset.

Correlation Value: The correlation between household size and total income is 0.14, indicating a weak positive relationship between the two variables. This suggests that in general, larger households tend to have slightly higher incomes, but the relationship is not strong.

```python
# Plotting the relationship between 'Total Number of Family Members' and 'Total Income'
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Total Number of Family Members', y='Total Income', data=newDf, alpha=0.6)

# Adding a regression line
sns.regplot(x='Total Number of Family Members', y='Total Income', data=newDf, scatter=False, color='red')

# Customize the plot
plt.title('Relationship Between Household Size and Total Household Income')
plt.xlabel('Total Number of Family Members')
plt.ylabel('Total Income')
plt.grid(True)
plt.tight_layout()

# Show the plot
plt.show()
```

### Conclusion
Larger households are generally associated with lower or moderate total household incomes based on the data, though there are exceptions where large households also have higher incomes. However, the correlation is weak, suggesting that other factors contribute more significantly to determining household income.
